```python
# Import Libraries
import cv2
from crop_stitcher import crop_stitcher
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
# Read Images
img = cv2.imread("examples/panorama.jpg")

# Convert BGR to RGB (cv2 reads images in BGR format)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

cropped, (x, y, w, h) = crop_stitcher(img, black_thresh=5)

# Display Cropped Image
plt.figure(figsize=[30, 10])
plt.imshow(cropped)
```

```python
# Optional: draw the detected rectangle (like the second image above)
img_copy = img.copy()
img_copy = cv2.rectangle(img_copy, (x, y), (x + w - 1, y + h - 1), (255, 0, 0), 4)

# Display Detected Rectangle
plt.figure(figsize=[30, 10])
plt.imshow(img_copy)
```
